## <b> Filling Missing Data (fillna): When you cannot afford to lose
## data, you can replace missing cells with alternative values using df.fillna() <b>


```python
df = pd.read_csv(r'C:\Users\USER\OneDrive\Desktop\Webfala\pandas\dirty_data.csv')
```

```python
df.info()
```

```python
# Constant value substitution: Fills every missing cell with a specific n
# Fill all NaNs with 0
# But this might not be suitable in all cases
df.fillna(0, inplace=True) 
df.info()
```

```python
# Fill with a default string (not really useful too) only for place where your values are strings

df['Calories'].fillna('Unknown')
df.info()
```

```python
# Statistical Imputation (Numerical columns):
# Replaces missing data with the column's mean, median, or mode.

# df['Calories'].fillna(df['Calories'].mean(), inplace=True) # old way

df['Calories'] = df['Calories'].fillna(df['Calories'].mean())
```

## Combining DataFrames

```python
# a. concat(): Combine along rows or columns
data1 =  {
 'Name': ['Rhoda', 'Jemilah', 'Rahmat', 'Esther'],
 'Age': [25, 30, 35, 10],
 'City': ['Ilorin', 'Lagos', 'Ibadan', 'Abuja']
}

df1 = pd.DataFrame(data1)
df1
```

```python
data2 = {
 'Name': ['Abiola', 'Nofisat', 'Salamat', 'Lola'],
 'Age': [20, 20, 35, 11],
 'City': ['Abuja', 'Akure', 'Mina', 'Kano']
}

df2 = pd.DataFrame(data2)
df2
```

```python
# Stack rows
# remember the concat we did earlier?
# remeber to ignore index

pd.concat([df1, df2], axis=0, ignore_index=True)
```

```python
# Side-by-side
# we don't need to ignore index here

pd.concat([df1, df2], axis=1)
```

```python
# merge(): SQL-style joins on columns
# Inner join by default

pd.merge(df1, df2, on='Age')

# combines df1 and df2 side-by-side by matching rows that share the same
# Includes only matching keys from both DataFrames. Rows with keys that d
```
